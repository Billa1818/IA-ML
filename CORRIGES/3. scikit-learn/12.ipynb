### Exercice 12 — Clustering hiérarchique

```python
import numpy as np

import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs

from sklearn.cluster import KMeans

from sklearn.metrics import adjusted_rand_score

from scipy.cluster.hierarchy import linkage, dendrogram, fcluster
```

### Generation du dataset

```python
df , y = make_blobs(n_samples=500, centers=4, n_features=2, cluster_std=1.4, random_state=10)


Z = linkage(df, method="ward")
```

### Dendrogramme

```python
plt.figure(figsize=(16, 7))

dendrogram(
    Z,

    leaf_rotation=90,

    leaf_font_size=6
    
)

plt.title(" Methode de Ward")

plt.xlabel("Observations")

plt.ylabel("Distance")
```

### Partition 2, 4 et 6 clusters

```python
clusters_2 = fcluster(

    Z,

    t=2,

    criterion="maxclust"

)

clusters_4 = fcluster(

    Z,

    t=4,

    criterion="maxclust"

)

clusters_6 = fcluster(

    Z,

    t=6,

    criterion="maxclust"
    
)
```

### Visualisation 

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].scatter(

    df[:, 0],

    df[:, 1],

    c=clusters_2,

    cmap="viridis"

)
axes[0].set_title("2 clusters")

axes[1].scatter(

    df[:, 0],

    df[:, 1],

    c=clusters_4,

    cmap="viridis"
)
axes[1].set_title("4 clusters")

axes[2].scatter(

    df[:, 0],

    df[:, 1],

    c=clusters_6,

    cmap="viridis"
)
axes[2].set_title("6 clusters")

for ax in axes:

    ax.set_xlabel("Variable 1")
    
    ax.set_ylabel("Variable 2")

plt.tight_layout()
```

### Kmean

```python

kmeans = KMeans(

    n_clusters=4,

    random_state=10,
    
    n_init=10
)

kmeans_labels = kmeans.fit_predict(df)
```

### Comparaison

```python
ari = adjusted_rand_score(

    kmeans_labels,
    
    clusters_4
)

ari
```
