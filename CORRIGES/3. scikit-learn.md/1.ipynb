## Exercice 1 — Nettoyage et encodage

```python
import numpy as np
import pandas as pd
from sklearn.impute import  SimpleImputer , KNNImputer
from sklearn.preprocessing import LabelEncoder , OneHotEncoder
np.random.seed(42)
from sklearn.compose import ColumnTransformer
```

### Création du dataframe

```python
df = pd.DataFrame({
    "age":np.random.normal(40,12,200).astype(int),
    "salaire_annuel": np.random.normal(45000,15000,200).round(0),
    "taille_cm": np.random.normal(172,9,200).astype(int),
    "niveau_education" : np.random.choice(["Collège", "Lycée", "Bac+2", "Bac+3", "Bac+5", "Doctorat"],200),
    "ville": np.random.choice(["Paris", "Lyon", "Marseille", "Toulouse", "Nice", "Nantes"],200),
    "genre": np.random.choice(["Homme", "Femme", "Autre"],200),
    "statut_marital": np.random.choice(["Célibataire", "Marié", "Divorcé", "Veuf"],200)

})
```

```python
df
```

### Suspression de valeur pour avoir des valeurs manquantes

```python
nan_counts = {
    "age": 15,
    "salaire_annuel": 20,
    "taille_cm": 10,
    "niveau_education": 12,
    "ville": 8,
    "genre": 5,
    "statut_marital": 10
}

for colonne, nombre in nan_counts.items():
    indices = np.random.choice(
        df.index,
        size=nombre,
        replace=False
    )
    df.loc[indices, colonne] = np.nan
```

### Remplacement des valeur numerique manquante par la mediane et valeur recurent


```python

colonne_numerique = ["age","salaire_annuel","taille_cm"]

colonne_categorielle = ["niveau_education","ville","genre","statut_matiral"]

imputer_num = SimpleImputer(strategy='mean')

imputer_cat = SimpleImputer(strategy="most_frequent")

df[colonne_numerique] = imputer_num.fit_transform(df[colonne_numerique])

df[colonne_categorielle] = imputer_cat.fit_transform(df[colonne_categorielle])

```

### Encodage du niveau d'education

```python
encoder_edu =  LabelEncoder()

df['niveau_education'] = encoder_edu.fit_transform(df['niveau_education'])
```

```python
other_encoder = OneHotEncoder(sparse_output=False).set_output(transform="pandas")

other_col = ['ville','genre','statut_marital']

new_other = other_encoder.fit_transform(df[other_col])

df = df.drop(columns=other_col)

df = df.join(new_other)
df
```
