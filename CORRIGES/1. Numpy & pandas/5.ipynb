# Exercice 5 — Algèbre linéaire

```python
import numpy as np
```

### Creation de deux matrice carré 

```python
mat1 = np.random.randint(1,9,(3,3))

mat1
```

```python
mat2 = np.random.randint(1,9,(3,3))

mat2
```

### Produit matricielle 

```python
mat = mat1 * mat2

mat 
```

### Calculons l'inverse 

```python
np.linalg.inv(mat)
```

### Le déterminant

```python
np.linalg.det(mat)
```

### Resolution de Ax = B

```python
A = [[2,1],[5,3]]

b = [8,13]

x = np.linalg.solve(A, b)

x
```

### Valeur et vecteur propre

```python
val_p , vec_p = np.linalg.eig(mat) 

val_p
```
