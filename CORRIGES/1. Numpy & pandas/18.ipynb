```python
import pandas as pd

import numpy as np
```

### Fixation du la graine de reproductibilité

```python
np.random.seed(42)
```

### Création du DataFrame

```python
df = pd.DataFrame({
    "x": np.random.uniform(-100, 100, 100_000),
    "y": np.random.uniform(-100, 100, 100_000)
})
```

### Calcul de la distance ecludienne

```python

X = df["x"].to_numpy()
Y = df["y"].to_numpy()

distances_numpy = np.sqrt(X**2 + Y**2)
```

### Ajout de la distance

```python
df["distance"] = distances_numpy
```

### Statistiques

```python
print(df["distance"].describe())
```
