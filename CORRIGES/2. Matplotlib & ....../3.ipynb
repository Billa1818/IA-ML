# Exercice 3 — Nuage de points et corrélation

```python
import numpy as np

import pandas as pd

import matplotlib.pyplot as plt
```

### Creation du dataframe

```python
data = {
    "age": np.random.randint(20, 61, 200),
    "salaire": np.random.randint(80000, 600000, 200),
    "categorie": np.random.choice(
        ["junior", "confirme", "senior"],
        200,
        p=[0.4, 0.35, 0.25]
    )
}

df = pd.DataFrame(data)
```

```python
df
```

### Creation du graphe

```python
fig , ax = plt.subplots(1,2,figsize=(16,9))

couleurs = {
    "junior": "blue",
    "confirme": "orange",
    "senior": "green"
}


ax[0].scatter(
    df["age"],
    df["salaire"],
    c=df["categorie"].map(couleurs)
)

ax[0].set_title("Salaire en fonction de l'age")

ax[0].set_xlabel("age")

ax[0].set_ylabel("Salaire")


a, b = np.polyfit(df["age"], df["salaire"], 1)

x = np.linspace(df["age"].min(), df["age"].max(), 100)

y = a * x + b


ax[1].scatter(
    df["age"],
    df["salaire"],
    color="steelblue",
    alpha=0.5,
    label="Individus"
)

ax[1].plot(
    x,
    y,
    color="red",
    linewidth=2,
    label="Tendance globale"
)

ax[1].set_title("Tendance globale age / salaire")

ax[1].set_xlabel("age")

ax[1].set_ylabel("Salaire")

ax[1].legend()

ax[1].grid(alpha=0.3)
```
