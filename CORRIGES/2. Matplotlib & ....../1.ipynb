### Exercice 1 — Graphique linéaire

```python
import matplotlib.pyplot as plt

import numpy as np 

import pandas as pd
```

### Generation du 90j  de ventes journaliere aleatoire

```python
df =  pd.DataFrame({

    "date" : pd.date_range(start="2026-01-01",periods=90,freq="D"),

    "produit_a" : np.random.randint(0,150,90),

    "produit_b" : np.random.randint(0,150,90),

    "produit_c" : np.random.randint(0,150,90),

})
```

```python
df
```

### Traçons leurs évolutions sur un même graphique avec une couleur distincte par produit

```python
fg ,  ax = plt.subplots()

ax.set_title("Evolution de vente des produit A, B et C")

ax.plot(df['date'],df['produit_a'],label="Produit A")

ax.plot(df['date'],df['produit_b'],label="Produit B")

ax.plot(df['date'],df['produit_c'],label="Produit C")

ax.set_xlabel('Date')

ax.set_ylabel('Quantité vendu')

ax.legend(loc='best')

ax.grid(True,linestyle = "-.",alpha=0.3)
```
