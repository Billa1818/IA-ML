### 2. House Prices — Advanced Regression

```python
import numpy as np

import pandas as pd

import seaborn as sns

import matplotlib.pyplot as plt
```

### Importation des données d'entrainement et de test

```python
df_train =  pd.read_csv("dataset/2/train.csv")

df_test =  pd.read_csv("dataset/2/test.csv")
```

```python
df_train
```

### Extraction de Y

```python
y = df_train["SalePrice"]

df_train = df_train.drop("SalePrice",axis=1)
```

### Supression des doublon

```python
X =  df_train.drop_duplicates()
```

### Verification des valeur manquante

```python
nan_val  = X.isna().sum()

fig , ax = plt.subplots(figsize=(30,10))

ax.bar(nan_val.index, nan_val.values)

ax.tick_params(axis="x", rotation=90)

ax.set_ylabel("Valeur NAN")

ax.set_title("Nombre de NAN")
```

### Etude des distributions des variables

```python
num_features = df_train.select_dtypes(include=np.number).columns

cat_features = df_train.select_dtypes(

    include=["object", "category"]

).columns

for col in num_features:

    fig, ax = plt.subplots(figsize=(8, 4))

    sns.histplot(

        data=df_train,

        x=col,

        kde=True,

        bins=30,
        
        ax=ax
    )

    ax.set_title(f"Distribution de {col}")
```
