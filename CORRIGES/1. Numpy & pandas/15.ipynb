# Exercice 15 — Génération et analyse de données

```python
import numpy as np

import pandas as pd
```

### Génère avec des outils numériques un jeu de données de 200 personnes avec un âge un salaire et un score

```python
age = np.random.randint(18,75,200)
```

```python
salaire = np.random.uniform(500,12000,200).round(2)
```

```python
score = np.random.rand(200).round(2)
```

### Creation du dataframe

```python
df = pd.DataFrame({
    "age" : age ,
    "salaire" : salaire ,
    "score" : score
})
```

```python
df
```

### Calcul des quartiles du salaire et ajout de colonne qui classe chaque personne selon son quartile

```python
df["salaire"].quantile([0.25, 0.50, 0.75])
```

```python
df["quartile"] = pd.qcut(
    df["salaire"],
    q=4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)
```

```python
df
```

### Calcul de correlation entre age et salaire

```python
df["age"].corr(df["salaire"])
```
