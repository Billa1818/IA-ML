# Exercice 13 — Séries temporelles

```python
import pandas as pd

import numpy as np
```

### Crée une série de 90 valeurs journalières aléatoires sur 3 mois

```python
dates = pd.date_range(start="2020-01-01",periods=90,freq="D")
```

```python
valeurs = np.random.randint(0,99,90)
```

```python
df = pd.DataFrame({

    "date" : dates,

    "valeur" : valeurs
})
```

```python
df
```

### Filtrages des données du mois de février.

```python
df[df['date'].dt.month==2]
```

### Calcule de la moyenne hebdomadaire.

```python
df.set_index("date").resample("W")["valeur"].mean()
```

### Ajout d'une colonne indiquant le jour de la semaine

```python
df["jours"] = df["date"].dt.day_name()
```

```python
df
```

### moyenne mobile sur 7j

```python
df["moyenne_mobile_7j"] = df["valeur"].rolling(window=7).mean()
```
