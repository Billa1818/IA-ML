# Exercice 11 — GroupBy et agrégations


```python
import pandas as pd
```

### Creation du dataframe

```python
data = {
    "region": [
        "Nord", "Sud", "Est", "Ouest", "Nord",
        "Sud", "Est", "Ouest", "Nord", "Sud"
    ],
    "vendeur": [
        "Alice", "Jean", "Marie", "Paul", "Lucas",
        "Emma", "Thomas", "Julie", "Pierre", "Sophie"
    ],
    "mois": [
        "Janvier", "Janvier", "Février", "Février", "Mars",
        "Mars", "Avril", "Avril", "Mai", "Mai"
    ],
    "montant": [
        2500, 3200, 1800, 4000, 3500,
        2800, 4200, 3100, 4500, 2900
    ]
}
```

```python
df = pd.DataFrame(data)
```

### Calcul du total des ventes par region

```python
df.groupby("region")["montant"].sum()
```

### Calcul de la moyenne par vendeur

```python
df.groupby("vendeur")["montant"].mean()
```

### Calcul du max par vendeur

```python
df.groupby("vendeur")["montant"].max()
```

### Moyenne et max par vendeur

```python
df.groupby("vendeur")["montant"].agg(["mean", "max"])
```

### Resumé par region

```python
df.groupby("vendeur")["montant"].agg(["sum", "mean", "count"]) 
```

### Creation du tableau croisé dynamique

```python
table = pd.pivot_table(
    df,
    values="montant",
    index="mois",
    columns="region",
    aggfunc="sum",
)
```

```python
table
```

```python
total_region = df.groupby("region")["montant"].transform("sum")

total_region
```

```python
df["Pourcentage_vente"] = (df["montant"] / total_region)*100
```
