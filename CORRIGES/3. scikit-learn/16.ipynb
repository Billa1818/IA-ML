# Exercice 16 — Métriques et choix du bon indicateur

```python
from sklearn.datasets import make_classification

from sklearn.model_selection import train_test_split

from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score , f1_score , roc_auc_score , precision_score , recall_score

import pandas as pd

import numpy as np
```

### Generation du dataset

```python
x , y  = make_classification(n_samples=2000, n_features=10, n_informative=5, weights=[0.95, 0.05], random_state=42)
```

### Separation du train et du test

```python
x_train , x_test , y_train , y_test = train_test_split(x,y , test_size=0.3)
```

### Entraînent d'un model  LogisticRegression par défaut

```python
model =  LogisticRegression()
model.fit(x_train,y_train)

pred =model.predict(x_test) 

model.score(x_test,y_test)
```

### Calcul du accuracy score

```python
acc_score = accuracy_score(y_test,pred)

acc_score
```

### Model Naif

```python
y_naif = np.zeros_like(y_test)

accuracy_naif = accuracy_score(y_test, y_naif)

accuracy_naif
```

### Comparaison des modèles

```python
resultats = []

resultats.append({
    "model": "Naif",
    "accuracy": accuracy_score(y_test, y_naif),
    "precision": precision_score(
        y_test,
        y_naif,
        pos_label=1,
        zero_division=0
    ),
    "rappel": recall_score(
        y_test,
        y_naif,
        pos_label=1,
        zero_division=0
    ),
    "f1-score": f1_score(
        y_test,
        y_naif,
        pos_label=1,
        zero_division=0
    ),
    "AUC-ROC": np.nan
})

y_proba = model.predict_proba(x_test)[:, 1]

resultats.append({
    "model": "Logistic Regression",
    "accuracy": accuracy_score(y_test, pred),
    "precision": precision_score(
        y_test,
        pred,
        pos_label=1,
        zero_division=0
    ),
    "rappel": recall_score(
        y_test,
        pred,
        pos_label=1,
        zero_division=0
    ),
    "f1-score": f1_score(
        y_test,
        pred,
        pos_label=1,
        zero_division=0
    ),
    "AUC-ROC": roc_auc_score(
        y_test,
        y_proba
    )
})

df_resultats = pd.DataFrame(resultats)

df_resultats
```

### 1. Détection de fraude bancaire — privilégier le rappel

Dans la détection de fraude bancaire, je privilégierais principalement le **rappel (recall)** de la classe « fraude ». Le faux négatif correspond ici à une transaction frauduleuse que le modèle considère comme normale. Son coût peut être très élevé pour la banque et pour le client, car la fraude n'est pas détectée et peut entraîner une perte financière importante. Un faux positif, au contraire, correspond à une transaction légitime signalée comme frauduleuse. Même s'il peut être gênant pour le client, il est généralement possible de demander une vérification supplémentaire ou de bloquer temporairement la transaction. Il est donc préférable de tolérer davantage de faux positifs afin de réduire le nombre de fraudes non détectées. Le **rappel** est ainsi une métrique essentielle, éventuellement complétée par la précision et le F1-score pour contrôler le nombre d'alertes inutiles.

### 2. Diagnostic médical — privilégier le rappel

Dans le diagnostic médical d'une maladie grave, je privilégierais également le **rappel** de la classe « malade ». Le faux négatif est particulièrement dangereux : il correspond à un patient malade que le modèle considère comme sain, ce qui peut retarder son diagnostic et sa prise en charge. Le coût humain potentiel est donc très important. À l'inverse, un faux positif correspond à un patient sain que le modèle considère comme potentiellement malade. Cela peut entraîner des examens complémentaires, du stress et des coûts supplémentaires, mais ces conséquences sont généralement moins graves qu'une maladie non détectée. Dans ce contexte, on cherchera donc à maximiser le **rappel**, tout en surveillant la précision afin d'éviter un nombre excessif de faux positifs.
