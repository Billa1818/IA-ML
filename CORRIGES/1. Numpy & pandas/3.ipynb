### Exercice 3 — Opérations et broadcasting

```python
import numpy as np
```

### Tableau de 5 entier

```python
tab =  np.arange(5)

tab
```

### Ajoute du scalaire sans boucle

```python
tab = tab + 2

tab
```

### Ajoute du carré par element

```python
tab =  tab ** 2

tab
```

### Normalisation entre 0 et 1

```python
tab  =  (tab - tab.min() ) / (tab.max() - tab.min())

tab
```

### Création d'une matrice 3x4  aléatoire

```python
mat_al = np.random.randint(0,9,(3,4))

mat_al
```

### Moyenne par ligne

```python
np.mean(mat_al,axis=0)
```

### Moyenne par colonne

```python
np.mean(mat_al,axis=1)
```

### Ecart-type global

```python
np.std(mat_al)
```
