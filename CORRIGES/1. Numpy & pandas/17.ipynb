# Exercice 17 — Calculs matriciels et réduction de dimension

```python
import pandas as pd

import numpy as np
```

### Creation du dataframe

```python
df = pd.DataFrame({
    "Mathematiques": [15, 12, 18, 10, 14],
    "Informatique": [17, 13, 19, 11, 15],
    "Physique": [14, 11, 16, 9, 13]
}, index=["Alice", "Bob", "Claire", "David", "Emma"])
```

### Conversion en matrice numerique

```python
X = df.to_numpy()

X
```

### Centrer autour de 0

```python
moyenne = np.mean(X, axis=0)

X_centre = X - moyenne

print(X_centre)
```

### Calcul de la matrice de covariance

```python
np.cov(X_centre, rowvar=False)
```

### Valeurs et vecteurs propres

```python
valeurs_propres, vecteurs_propres = np.linalg.eig(np.cov(X_centre, rowvar=False))
```

```python
valeurs_propres
```

```python
vecteurs_propres
```

### Projet les données sur les deux axes principaux et reconvertis le résultat en DataFrame avec des colonnes nommées PC1 et PC2.

```python
indices = np.argsort(valeurs_propres)[::-1]

valeurs_propres = valeurs_propres[indices]

vecteurs_propres = vecteurs_propres[:, indices]

axes_principaux = vecteurs_propres[:, :2]

print(axes_principaux)

projection = X_centre @ axes_principaux

print(projection)
```

```python
df_pca = pd.DataFrame(
    projection,
    columns=["PC1", "PC2"],
    index=df.index
)

print(df_pca)
```
