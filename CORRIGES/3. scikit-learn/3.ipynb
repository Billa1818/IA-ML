# Exercice 3 — Séparation train / test et validation croisée

```python
from sklearn.datasets import make_classification

from sklearn.model_selection import train_test_split , cross_val_score

from sklearn.ensemble import RandomForestClassifier

from sklearn.model_selection import StratifiedKFold

import pandas as pd
```

```python
x , y = make_classification(
    
    n_samples=1000,

    n_features=10,

    n_informative=6,

    n_classes=3,

    weights=[0.6, 0.3, 0.1],

    random_state=42

)
```

###  Séparation en train/test

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size=0.25, stratify=y, random_state=42)
```

### Affichage la proportion de chaque classe dans y, y_train et y_test, et vérifie qu'elles sont quasi identiques (écart < 2 points de %).

```python
df = pd.DataFrame({
    "y" : y,
})

df['y'].value_counts(normalize=True) *100
```

```python
df = pd.DataFrame({
    "y_train" : y_train,
})

df['y_train'].value_counts(normalize=True)
```

```python
df = pd.DataFrame({
    "y_test" : y_test,
})

df['y_test'].value_counts(normalize=True)
```

### Entraînement d'un RandomForestClassifier

```python
model =  RandomForestClassifier(random_state=42)

model.fit(x_train,y_train)

model.score(x_train,y_train)

model.score(x_test, y_test)
```

### Mis en place une validation croisée en 5 blocs

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)


score = cross_val_score(model,x_train,y_train,cv=cv,scoring="f1_macro")


```

### Affiche le score moyen et l'écart-type des 5 scores obtenus.

```python
score.mean()
```

```python
score.std()
```
