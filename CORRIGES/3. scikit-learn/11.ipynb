# Exercice 11 — K-Means et choix du nombre de clusters

```python
import pandas as pd

from sklearn.datasets import make_blobs 

from sklearn.cluster import KMeans

import matplotlib.pyplot as plt
```

### Generation du dataset

```python
df , y = make_blobs(n_samples=500, centers=4, n_features=2, cluster_std=1.4, random_state=10)

df
```

### Application de KMeans avec k -> 1 a 10

```python
fig , ax =  plt.subplots(1,10,figsize=(16,5))

for k in range(1,11):

    model =  KMeans(n_clusters=k)

    model.fit(df)

    pred = model.predict(df)

    ax[k-1].scatter(df[:,0],df[:,1],c=pred)

    ax[k-1].set_xlabel(f"K ={k}")
```

### Réentraînement du model K-Means avec le k = 4

```python
fig , ax =  plt.subplots(figsize=(10,5))

model =  KMeans(n_clusters=4)

model.fit(df)

pred = model.predict(df)

ax.scatter(df[:,0],df[:,1],c=pred)

ax.scatter(model.cluster_centers_[:,0], model.cluster_centers_[:,1],c='red')

ax.set_xlabel("K = 4")
```

### Regroupe les points par cluster 

```python
df_cluster = pd.DataFrame(df , columns=["x","y"])

df_cluster["cluster"] =  pred

df_cluster.groupby('cluster').mean()
```
