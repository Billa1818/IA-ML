# Exercice 7 — Performance et vectorisation


```python
import numpy as np

from datetime import datetime 
```

### Generation d'un million d'element

```python
tab = np.random.randint(0,9,1000000)
```

### Comparaison de somme avec boucle et vectorisation

Boucle

```python
start =  datetime.now()

sum = 0

for i in tab :

    sum =  sum + tab[i]

end = datetime.now()


exec_temp =  end - start 

exec_temp.total_seconds()
```

Vectorisation

```python
start =  datetime.now()

tab.sum()

end = datetime.now()


exec_temp =  end - start 

exec_temp.total_seconds()
```
