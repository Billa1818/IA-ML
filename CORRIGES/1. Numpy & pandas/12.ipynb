# Exercice 12 — Jointures

```python
import pandas as pd
```

### Creation des dataframe

```python
data_clients = {
    "id_client": [1, 2, 3, 4, 5],
    "nom": ["Alice", "Jean", "Marie", "Paul", "Sophie"]
}

data_commandes = {
    "id_commande": [101, 102, 103, 104, 105, 106, 107, 108],
    "id_client": [1, 2, 1, 3, 5, 2, 4, 3],
    "montant": [250, 450, 120, 800, 350, 150, 600, 200]
}

clients = pd.DataFrame(data_clients)

commandes =  pd.DataFrame(data_commandes)
```

```python
clients
```

```python
commandes
```

### Jointure pour garder les client ayant au moin une commande

```python
dataf = pd.merge(clients , commandes , on="id_client" , how="inner")

dataf
```

### Jointure pour garder tous les clients même sans commande

```python
pd.merge(clients , commandes , on="id_client" , how="left")
```

### Une dernière pour tout conserver des deux côtés

```python
pd.merge(clients , commandes , on="id_client" , how="outer")
```

 ### Concatenation de deux dataframe 

```python
cl1 = pd.DataFrame({
    "id": [1, 2, 3, 4, 5],
    "nom": ["Alice", "Bob", "Charlie", "David", "Emma"],
    "age": [22, 25, 30, 28, 24],
    "ville": ["Cotonou", "Parakou", "Porto-Novo", "Abomey-Calavi", "Bohicon"]
})

cl2 = pd.DataFrame({
    "id": [3, 4, 5, 6, 7],
    "nom": ["Charlie", "David", "Emma", "Frank", "Grace"],
    "age": [30, 28, 24, 35, 27],
    "ville": ["Porto-Novo", "Abomey-Calavi", "Bohicon", "Cotonou", "Parakou"]
})
```

```python
df =  pd.concat([cl1, cl2], axis=0)
```

### Total commander par client

```python
total = commandes.groupby(
    "id_client",
    as_index=False
)["montant"].sum()

total = total.rename(columns={
    "montant": "total_commandes"
})
```
