# Exercice 9 — Comparaisons par catégories

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

### Dataset

```python
mois = [
    "Janvier", "Fevrier", "Mars", "Avril",
    "Mai", "Juin", "Juillet", "Aout",
    "Septembre", "Octobre", "Novembre", "Decembre"
]

regions = ["Nord", "Sud", "Centre"]

df_ventes = pd.DataFrame({
    "Mois": mois * 3,
    "Region": np.repeat(regions, 12),
    "Ventes": np.concatenate([
        np.random.randint(1000, 3000, 12),
        np.random.randint(1500, 3500, 12),
        np.random.randint(2000, 4000, 12)
    ])
})

df_ventes.head()
```

```python
plt.figure(figsize=(15, 7))

sns.barplot(
    data=df_ventes,
    x="Mois",
    y="Ventes",
    hue="Region",
    palette="Set2"
)

plt.title("Comparaison des ventes mensuelles par région")
plt.xlabel("Mois")
plt.ylabel("Montant des ventes")
plt.xticks(rotation=45)
plt.legend(title="Régions")
plt.grid(axis="y", alpha=0.3)
```
