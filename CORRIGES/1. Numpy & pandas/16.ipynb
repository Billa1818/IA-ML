# Exercice 16 — Transformations en masse

```python
import numpy as np

import pandas as pd
```

### Génère avec des outils numériques un jeu de données de 200 personnes avec un âge un salaire et un score


```python
age = np.random.randint(18,75,200)
```

```python
salaire = np.random.uniform(500,12000,200).round(2)
```

```python
score = np.random.rand(200).round(2)
```

### Creation du dataframe

```python
df = pd.DataFrame({
    "age" : age ,
    "salaire" : salaire ,
    "score" : score
})
```

### Création d'une colonne booléenne indiquant si la personne est senior

```python
def cat(val):
    if val >=30:
        return 1
    else : return 0

df["categorie"] = df['age'].apply(cat) 

df
```

### Plafonisation des salaire

```python
df["salaire"]  = df["salaire"].clip(lower=1000,upper=10000)

df
```

### Categorisation bassées sur le score

```python
seuil = [0.3,0.7,0.9,1]

label = ["faible","moyen","eleve"]

df["cat_score"] = pd.cut(df['score'],bins=seuil,labels=label)

df
```
