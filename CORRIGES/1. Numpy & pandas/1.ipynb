# Exercice 1 — Création de tableaux

```python
import numpy as np
```

### Tableau d'entier de 1 a 10

```python
tab1 = np.arange(10)

tab1
```

### Tableau cinq zéros en virgule flottante

```python
tab2 = np.zeros(5)

tab2
```

### Tableau de six valeurs régulièrement espacées entre 0 et 1

```python
tab3 = np.linspace(0,1,6)

tab3
```

### Tableau d'une matrice identité 4×4

```python
tab4 = np.eye(4)

tab4
```
