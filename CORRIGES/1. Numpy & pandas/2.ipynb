# Exercice 2 — Indexation et slicing

```python
import numpy as np
```

### Création d'une matrice 3x3

```python
mat = np.arange(10,100,10).reshape(3,3)

mat
```

### Extration de la 2eme ligne

```python
ligne2 = mat[1,:]

ligne2
```

### Extraction de la 3eme colonne

```python
col3 = mat[:,2]

col3
```

### Sous matrice 2x2 en bas a droite

```python
s_mat = mat[1:,1:]

s_mat
```

###  Tous les éléments supérieurs à 45

```python
sup45 = mat[mat > 45]

sup45
```

### Remplacer tout les element paire par 0

```python
mat[ mat % 2 == 0 ] = 0

mat
```
