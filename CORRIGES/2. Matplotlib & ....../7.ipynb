# Exercice 7 — Relations entre variables

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df =  pd.DataFrame({

    "age" : np.random.randint(18,80,200),

    "salaire" : np.random.uniform(1000,20000,200),

    "score" : np.random.rand(200),

    "anciennete" : np.random.randint(1,30,200),

    "categorie" : np.random.choice(["junior","pro","senior"],200)

})

df
```

```python
sns.pairplot(
    data=df,
    vars=["age", "salaire", "score", "anciennete"],
    hue="categorie",
    palette="Set1",
    diag_kind="kde",
    plot_kws={"alpha": 0.6}
)

plt.suptitle(
    "Relations entre les variables numériques",
    y=1.02
)
```
