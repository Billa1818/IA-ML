# Exercice 2 — Barres et camembert

```python
import pandas as pd

import matplotlib.pyplot as plt
```

```python
data = {
    "region": ["Nord", "Sud", "Est", "Ouest", "Centre"],
    "ventes_totales": [125000, 185000, 95000, 210000, 145000]
}

df = pd.DataFrame(data)

df
```

### Traçons

```python
fg ,  ax = plt.subplots(2,figsize=(16,9))

ax[0].barh(df["region"],df["ventes_totales"])

ax[0].set_title("Diagramme en barre des montant sur differentes periodes")

ax[1].set_title("Diagramme circulaire des montant sur differentes periodes")

colors = ["red" if val == max else "lightgray" for val in df["ventes_totales"]]

explode = [0.08 if val == max else 0 for val in df["ventes_totales"]]

ax[1].pie(df["ventes_totales"],labels=df["region"],colors=colors,autopct="%1.1f%%",explode =explode)


```
