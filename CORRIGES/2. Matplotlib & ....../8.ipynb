# Exercice 8 — Heatmap de corrélation

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

### Creation du dataset

```python
df =  pd.DataFrame({

    "age" : np.random.randint(18,80,200),

    "salaire" : np.random.uniform(1000,20000,200),

    "score" : np.random.rand(200),

    "anciennete" : np.random.randint(1,30,200),

    "categorie" : np.random.choice(["junior","pro","senior"],200)

})

df
```

```python
correlation = df[
    ["age", "salaire", "score", "anciennete"]
].corr()

print(correlation)
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5,
    vmin=-1,
    vmax=1,
    center=0
)

plt.title("Matrice de corrélation des variables")
```
