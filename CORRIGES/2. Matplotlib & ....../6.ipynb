# Exercice 6 — Distribution et densité

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

### Generation du dataset

```python
df =  pd.DataFrame({

    "Mathematique" : np.random.normal(15,5,300).round(0),

    "Francais" : np.random.normal(12,3,300).round(0),

    "Anglais" : np.random.normal(13,6,300).round(0),

    "Chimie" : np.random.normal(10,4,300).round(0)

})

df
```

```python
plt.figure(figsize=(16,9))


for i in df.columns:

    sns.kdeplot( data=df, x=i)
```

```python
for i in df.columns:

    sns.violinplot( data=df,inner="box")
```
