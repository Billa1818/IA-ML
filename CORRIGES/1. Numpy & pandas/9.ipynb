# Exercice 9 — Sélection et filtrage

```python
import pandas as pd
```

Creation du dataframe

```python
data = {
    "nom": ["Billa", "Alice", "Jean", "Marie", "Paul"],
    "age": [22, 25, 30, 28, 35],
    "ville": ["Paris", "Lyon", "Marseille", "Toulouse", "Bordeaux"],
    "salaire": [3200, 2800, 2600, 2900, 3100]
}

df =  pd.DataFrame(data)
```

### Selection des colonne nom et salaire

```python
new_df = df[["nom","salaire"]]
```

### Filtrage de personne de plus de 29 ans

```python
new_df[df['age']>29]
```

### Filtrage des personne habitant a paris ou a lyon

```python
new_df[(df["ville"]=="Paris") | (df["ville"]=="Lyon")]
```

Personne ayant un salaire entre 40000 et 50000

```python
new_df[(df["salaire"]>=40000) & (df["salaire"]<=50000) ]
```
