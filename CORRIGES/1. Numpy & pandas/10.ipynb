# Exercice 10 — Données manquantes 

```python
import pandas as pd
```

### Creation du dataframe du produit

```python
data = {
    "produit": [None, "Smartphone", "Tablette", "Clavier", "Souris"],
    "prix": [850, None, 350, 45, None],
    "quantite": [10, 15, 13, 30, 50],
    "categorie": ["Informatique", "Téléphonie", None, "Accessoires", "Accessoires"],
    "note": [4.5, 4.8, None, 4.2, None]
}
```

```python
df = pd.DataFrame(data)
```

### Indentification des celluled avec des valeurs manquantes

```python
df.isna()
```

### Pour le prix remplacement des valeurs manquantes par le moyenne

```python
df['prix'] = df['prix'].fillna(df['prix'].mean())
```

```python
df
```

### Pour les categories remplacement des valeurs manquantes par "inconnu"

```python
df['categorie'] = df['categorie'].fillna("inconnu")
```

```python
df
```

### Suspression des ligne avec nom du produit absent

```python
df.dropna(subset=['produit'])
```

### Conversion des valeur de la colonne stock en entier

```python
df["quantite"] = df["quantite"].astype("int64")
```
