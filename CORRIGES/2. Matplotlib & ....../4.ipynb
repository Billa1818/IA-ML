# Exercice 4 — Histogramme et distribution

```python
import numpy as np

import pandas as pd

import matplotlib.pyplot as plt

from scipy.stats import gaussian_kde
```

### Generation de 1000 salaire selon une distribution normal

```python
salaire = np.random.normal(loc=2500,scale=500,size=1000)
```

```python
df = pd.DataFrame({
    "salaire" : salaire
})
```

```python
df
```

### Traçons l'histogramme avec une courbe de densité

```python
fig , ax = plt.subplots(1,2,figsize=(16,9))

ax[0].set_title('Histogramme avec la courbe de densité')

ax[0].hist(df["salaire"],density=True)

densite = gaussian_kde(df["salaire"])

x = np.linspace(df["salaire"].min(),df["salaire"].max(),200)

ax[0].plot(
    x,
    densite(x),
    color="orange",
    linewidth=2,
    label="Courbe de densité"
)

moyenne =  df["salaire"].mean()

mediane = df["salaire"].median()

ax[0].axvline(
    moyenne,
    linestyle="-.",
    color = "red"
)

ax[0].axvline(
    mediane,
    linestyle="--",
    color = "blue"
)

ax[1].set_title("boite a moustache")

ax[1].boxplot(df["salaire"])
```
